```python
import os
import shutil

# 1. Caminho principal da pasta no Drive
caminho_principal = "/content/drive/MyDrive/Episodios_Anime/"

print("--- Começando a desfazer as subpastas ---")

# 2. O robô olha tudo o que está dentro da pasta principal
for item in os.listdir(caminho_principal):

    # Monta o caminho completo do item encontrado
    caminho_item = os.path.join(caminho_principal, item)

    # Verifica se o item é realmente uma pasta (e não um arquivo solto)
    if os.path.isdir(caminho_item):

        # 3. Entra na subpasta e olha os arquivos que estão dentro dela
        for arquivo in os.listdir(caminho_item):
            caminho_arquivo_atual = os.path.join(caminho_item, arquivo)
            caminho_destino_final = os.path.join(caminho_principal, arquivo)

            # Move o arquivo de volta para a pasta principal (raiz)
            shutil.move(caminno_arquivo_atual if 'caminno_arquivo_atual' in locals() else caminho_arquivo_atual, caminho_destino_final)
            print(f"Retirado da pasta e movido para a raiz: {arquivo}")

        # 4. Apaga a subpasta que agora ficou vazia
        os.rmdir(caminho_item)
        print(f"Pasta vazia removida: {item}")

print("\n--- Pronto! Todos os arquivos estão soltos na raiz, prontos para o TeraBox. ---")

```

# AUTOMAÇÃO LEGENDAS GPUT4- WHISPER

```python
import os  # Para checar se o arquivo existe

# Aqui você coloca apenas a lista dos números dos episódios que quer traduzir
numeros_episodios = [51, 52, 53, 54, 55]

print(f"--- Iniciando a tradução do lote: {numeros_episodios} ---")

for numero_episodio in numeros_episodios:

    print(f"\n--- Processando o Episódio {numero_episodio} ---")

    # 1. Definimos onde está o vídeo (fonte) e onde queremos a legenda (resultado)
    caminho_video = f"/content/drive/MyDrive/Episodios_Anime/episodio_{numero_episodio:02d}.mp4"
    caminho_legenda = f"/content/drive/MyDrive/Episodios_Anime/episodio_{numero_episodio:02d}.srt"

    # 2. Executamos apenas a tradução (Whisper) usando o arquivo de vídeo
    # O Whisper é inteligente: ele consegue extrair o áudio do vídeo sozinho!
    !whisper "{caminho_video}" --model medium --task translate --output_format srt --output_dir "/content/drive/MyDrive/Episodios_Anime"

    # 3. A CONDIÇÃO DE SEGURANÇA (O "Guardião da Porta")
    if os.path.exists(caminho_legenda):
        print(f"Episódio {numero_episodio} traduzido e legenda salva com sucesso!")
    else:
        print(f"ATENÇÃO: O Episódio {numero_episodio} falhou! O arquivo de legenda não foi gerado.")

print("\nProcesso de tradução do lote finalizado!")
```
